<h4>Task 1 </h4>

```python
email = "Amit_ml@gmail.edu"
if  email.count('@') == 1  and '.' in email[email.find('@'):]:
    print("valid email")
else:
    print("invalid email")
```

```python
n = email.find('@')
l = email.find('.')
print(f"Username : {email[:n]}")
print(f"Domain : {email[n+1:l]}")

```

```python
if email.endswith(".com"):
    print("Commercial Domain")
elif email.endswith(".edu"):
    print("Educational Domain")
else:
    print("Other Domain")

```
