<h3>Task one </h3>

```python
import numpy as np
def array_factory(mode :str , shape:tuple , value:int | float | None ):
    try:
        if mode == "zeros":
            return np.zeros(shape)

        elif mode == "ones":
            return np.ones(shape)

        elif mode == "full":
            if value is None:
                raise ValueError("value is required for 'full' mode")
            return np.full(shape,value)

        elif mode == "identity":
            if len(shape) != 2 or shape[0] != shape[1]:
                raise ValueError(
                        "Identity matrix must have a square shape"
                    )
            return np.eye(shape[0])
        
        else:
            raise ValueError(f"Unknown mode: {mode}")

    except (TypeError, ValueError) as e:
        raise ValueError(f"Invalid array creation: {e}") from e


array_factory("full" , (4,2), 5 )

```

<h3>Task two </h3>

```python
import numpy as np
def secure_reshape_and_stack(data1:list[int],data2:list[list],new_shape:tuple):
    try:
        arr1 = np.array(data1)
        arr2 = np.array(data2)

        reshaped_array = arr1.reshape(new_shape)

        combined_dataset = np.vstack((reshaped_array,arr2))

        return combined_dataset

    except ValueError as e:
        raise ValueError(f"error : {e}")

secure_reshape_and_stack([2,1,4,4],[[1,2],[2,2]],(2,2))
```
